```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.metrics import silhouette_score

```

```python
df = pd.read_csv('/content/data.csv', encoding='latin1')

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
print(df.isnull().sum())
```

```python
print(df.columns.tolist())
```

```python
print("Duplicates before cleaning:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after cleaning:", df.duplicated().sum())
```

```python
print("Missing values before cleaning:")
print(df.isnull().sum())

df = df.dropna(subset=['CustomerID'])

print("\nMissing Customer IDs after cleaning:",
      df['CustomerID'].isnull().sum())
```

```python
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]

df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)]

print("Dataset shape after cleaning:", df.shape)
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

print(df['InvoiceDate'].dtype)
```

```python
import datetime as dt

reference_date = df['InvoiceDate'].max() + dt.timedelta(days=1)

print("Reference Date:", reference_date)
```

```python
df['TotalAmount'] = df['Quantity'] * df['UnitPrice']

df.head()
```

```python
rfm = df.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (reference_date - x.max()).days,
    'InvoiceNo': 'nunique',
    'TotalAmount': 'sum'
})

rfm.columns = ['Recency', 'Frequency', 'Monetary']

rfm.head()
```

```python
print("RFM Dataset Shape:", rfm.shape)

rfm.describe()
```

```python
print("Skewness before transformation:")
print(rfm.skew())

rfm.hist(figsize=(12, 5), bins=30)
plt.tight_layout()
plt.show()
```

```python
rfm_log = np.log1p(rfm)

print("Skewness after transformation:")
print(rfm_log.skew())
```

```python
scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(rfm_log)

print("Scaled Data Shape:", rfm_scaled.shape)
```

```python
wcss = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(rfm_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), wcss, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS')
plt.title('Elbow Method')
plt.show()
```

```python
for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(rfm_scaled)

    score = silhouette_score(rfm_scaled, labels)

    print(f"K = {k}, Silhouette Score = {score:.4f}")
```

```python
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)

rfm['KMeans_Cluster'] = kmeans.fit_predict(rfm_scaled)

print(rfm['KMeans_Cluster'].value_counts())
```

```python
cluster_profile = rfm.groupby('KMeans_Cluster')[['Recency', 'Frequency', 'Monetary']].mean()

print(cluster_profile.round(2))
```

```python
pca = PCA(n_components=2)

rfm_pca = pca.fit_transform(rfm_scaled)

print("PCA Shape:", rfm_pca.shape)
print("Explained Variance:", pca.explained_variance_ratio_)
print("Total Variance:", pca.explained_variance_ratio_.sum())
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    rfm_pca[:, 0],
    rfm_pca[:, 1],
    c=rfm['KMeans_Cluster'],
    cmap='viridis',
    alpha=0.6
)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('K-Means Clusters using PCA')
plt.colorbar(label='Cluster')
plt.show()
```

```python
dbscan = DBSCAN(eps=0.5, min_samples=5)

dbscan_labels = dbscan.fit_predict(rfm_scaled)

rfm['DBSCAN_Cluster'] = dbscan_labels

print("DBSCAN Cluster Distribution:")
print(rfm['DBSCAN_Cluster'].value_counts().sort_index())
```

```python
dbscan_profile = rfm[rfm['DBSCAN_Cluster'] != -1].groupby('DBSCAN_Cluster')[['Recency', 'Frequency', 'Monetary']].mean()

print(dbscan_profile.round(2))
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    rfm_pca[:, 0],
    rfm_pca[:, 1],
    c=rfm['DBSCAN_Cluster'],
    cmap='plasma',
    alpha=0.6
)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('DBSCAN Clusters using PCA')
plt.colorbar(label='Cluster')
plt.show()
```

```python
tsne = TSNE(
    n_components=2,
    random_state=42,
    perplexity=30,
    init='pca',
    learning_rate='auto'
)

rfm_tsne = tsne.fit_transform(rfm_scaled)

print("t-SNE Shape:", rfm_tsne.shape)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    rfm_tsne[:, 0],
    rfm_tsne[:, 1],
    c=rfm['KMeans_Cluster'],
    cmap='viridis',
    alpha=0.6
)

plt.xlabel('t-SNE Component 1')
plt.ylabel('t-SNE Component 2')
plt.title('K-Means Clusters using t-SNE')
plt.colorbar(label='Cluster')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    rfm_tsne[:, 0],
    rfm_tsne[:, 1],
    c=rfm['DBSCAN_Cluster'],
    cmap='plasma',
    alpha=0.6
)

plt.xlabel('t-SNE Component 1')
plt.ylabel('t-SNE Component 2')
plt.title('DBSCAN Clusters using t-SNE')
plt.colorbar(label='Cluster')
plt.show()
```

```python
comparison = pd.DataFrame({
    'Algorithm': ['K-Means', 'DBSCAN'],
    'Number of Clusters': [
        rfm['KMeans_Cluster'].nunique(),
        len(set(dbscan_labels) - {-1})
    ],
    'Noise Points': [
        0,
        (rfm['DBSCAN_Cluster'] == -1).sum()
    ],
    'Customers': [
        len(rfm),
        (rfm['DBSCAN_Cluster'] != -1).sum()
    ]
})

print(comparison)
```

```python
print("K-MEANS CLUSTER PROFILE")
print(cluster_profile.round(2))

print("\nDBSCAN CLUSTER PROFILE")
print(dbscan_profile.round(2))

print("\nCUSTOMER DISTRIBUTION")
print("Total Customers:", len(rfm))
print("K-Means Clusters:", rfm['KMeans_Cluster'].value_counts().to_dict())
print("DBSCAN Clusters:", rfm['DBSCAN_Cluster'].value_counts().to_dict())
```

```python
personas = []

for cluster, row in cluster_profile.iterrows():

    if row['Recency'] < 40 and row['Frequency'] >= 5 and row['Monetary'] >= 2000:
        persona = "Champions / High-Value Customers"
        strategy = "VIP rewards, loyalty benefits and exclusive offers"

    elif row['Recency'] > 100 and row['Frequency'] < 3:
        persona = "At-Risk / Low-Engagement Customers"
        strategy = "Win-back campaigns, discounts and personalized reminders"

    elif row['Frequency'] >= 5:
        persona = "Loyal Regular Customers"
        strategy = "Membership benefits and cross-selling"

    else:
        persona = "Occasional Customers"
        strategy = "Product recommendations and repeat-purchase offers"

    personas.append({
        'Cluster': cluster,
        'Persona': persona,
        'Recency': round(row['Recency'], 2),
        'Frequency': round(row['Frequency'], 2),
        'Monetary': round(row['Monetary'], 2),
        'Marketing Strategy': strategy
    })

persona_df = pd.DataFrame(personas)

print(persona_df.to_string(index=False))
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    persona_df['Persona'],
    rfm.groupby('KMeans_Cluster').size().values
)

plt.xlabel('Customer Persona')
plt.ylabel('Number of Customers')
plt.title('Customer Distribution by Marketing Persona')
plt.xticks(rotation=15, ha='right')
plt.tight_layout()
plt.show()
```
